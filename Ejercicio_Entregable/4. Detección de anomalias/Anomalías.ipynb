### Detección de Fraude con Tarjetas de Crédito

**Objetivo**: Comparar diferentes técnicas de detección de anomalías para identificar transacciones fraudulentas de tarjetas de crédito.

#### Descripción del Conjunto de Datos
- **Conjunto de Datos**: "Credit Card Fraud Detection" de Kaggle.
- **Características**: Incluye características numéricas que son el resultado de una transformación PCA. Las únicas características que no han sido transformadas con PCA son 'Time' y 'Amount'.
- **Tarea**: Detectar casos de fraude a partir de transacciones normales.

#### Estructura del Ejercicio en Jupyter Notebook

1. **Introducción y Carga de Datos**
   - Presentación del problema de detección de fraudes y carga del conjunto de datos.
   - Análisis exploratorio inicial para entender la naturaleza de los datos.

2. **Preprocesamiento de Datos**
   - Normalización o estandarización de las características si es necesario.
   - División del conjunto de datos en conjuntos de entrenamiento y prueba.

3. **Aplicación de Técnicas de Detección de Anomalías**
   - Implementación y comparación de varias técnicas, como Isolation Forest, DBSCAN, y SVM de una clase (One-Class SVM).


4. **Evaluación y Comparación de Modelos**
   -   Análisis de los resultados y elección de métricas apropiadas para la evaluación (por ejemplo, precisión, recall, puntuación F1, área bajo la curva ROC).
   - Comparación de los modelos en términos de eficacia en la detección de fraudes.

5. **Conclusiones**
   - Discusión sobre las ventajas y desventajas de cada técnica.
   - Reflexiones sobre la importancia de la detección de anomalías en el contexto de seguridad financiera.

#### Tareas para los Estudiantes

- Analizar los resultados y discutir qué modelo(s) funcionan mejor y por qué.
- Investigar cómo la desproporción en la distribución de clases afecta la detección de anomalías y explorar posibles soluciones.
