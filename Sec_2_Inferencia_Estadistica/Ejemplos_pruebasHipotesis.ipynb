

### Ejemplo 1: Intervalo de Confianza para la Media

**Contexto**: Supón que trabajas para una empresa que vende productos en línea, y estás interesado en estimar el gasto promedio de los clientes en su primer pedido. Has recopilado una muestra de 30 pedidos recientes de clientes nuevos, y has calculado un gasto promedio de \$75 con una desviación estándar de \$15.

**Tarea**: Calcula un intervalo de confianza del 95% para el gasto promedio de los clientes en su primer pedido.

**Solución**:



```python

import scipy.stats as stats
import numpy as np

# Datos de la muestra
media_muestra = 75
desviacion_estandar = 15
n = 30

# Nivel de confianza
alpha = 0.05

# Grados de libertad
df = n - 1

# Estadístico t crítico
t_critico = stats.t.ppf(1 - alpha/2, df)

# Margen de error
margen_error = t_critico * (desviacion_estandar / np.sqrt(n))

# Intervalo de confianza
intervalo_confianza = (media_muestra - margen_error, media_muestra + margen_error)
intervalo_confianza

```


### Ejemplo 2: Prueba de Hipótesis para Diferencia de Medias

**Contexto**: Imagina que tu empresa ha implementado un nuevo diseño en la página de producto, y quieres saber si este cambio ha afectado al tiempo que los usuarios pasan en ella. Para esto, has recopilado los tiempos de permanencia de una muestra de usuarios antes y después del cambio. Tu hipótesis nula es que el tiempo medio de permanencia no ha cambiado, mientras que tu hipótesis alternativa es que el tiempo medio de permanencia ha aumentado.

```python
# Tiempos de permanencia antes y después del cambio
tiempos_antes = [120, 115, 130, 140, 125, 135, 110, 120, 125, 130]
tiempos_despues = [130, 120, 140, 145, 130, 138, 120, 125, 135, 140]

# Realizar la prueba t para muestras emparejadas
t_stat, p_valor = stats.ttest_rel(tiempos_despues, tiempos_antes, alternative='greater')

# Mostrar los resultados
t_stat, p_valor
```

La hipótesis nula se rechaza si el p-valor es menor que el nivel de significancia (por ejemplo, 0.05).



### Ejemplo 3: Prueba de Hipótesis para la Proporción

**Contexto**: Supón que tu empresa ha lanzado una nueva campaña de marketing y quieres saber si la tasa de conversión (la proporción de visitantes que realizan una compra) ha mejorado. Antes de la campaña, la tasa de conversión era del 10%. Después de la campaña, de 200 visitantes, 30 realizaron una compra.

**Tarea**: Realiza una prueba de hipótesis para determinar si la tasa de conversión ha mejorado significativamente.

**Solución**:



```python
import statsmodels.api as sm

# Tasa de conversión antes de la campaña
p0 = 0.10

# Resultados después de la campaña
compras = 30
visitantes = 200

# Realizar la prueba z para una proporción
z_stat, p_valor = sm.stats.proportions_ztest(count=compras, nobs=visitantes, value=p0, alternative='larger')

# Mostrar los resultados
z_stat, p_valor

```


Nuevamente, la hipótesis nula se rechaza si el p-valor es menor que el nivel de significancia.


### Ejemplo 4: Prueba de Hipótesis para Diferencia de Proporciones

**Contexto**: Imagina que trabajas para una empresa de software y has lanzado dos versiones de una nueva característica en tu aplicación. Quieres comparar las tasas de adopción de ambas versiones. En la versión A, 50 de 200 usuarios adoptaron la nueva característica, mientras que en la versión B, 30 de 150 usuarios la adoptaron.

**Tarea**: Realiza una prueba de hipótesis para determinar si existe una diferencia significativa en las tasas de adopción entre las dos versiones.

**Solución**:


```python
import statsmodels.api as sm

# Tasas de adopción
adopciones_A = 50
usuarios_A = 200
adopciones_B = 30
usuarios_B = 150

# Realizar la prueba z para la diferencia de proporciones
z_stat, p_valor = sm.stats.proportions_ztest([adopciones_A, adopciones_B], [usuarios_A, usuarios_B])

# Mostrar los resultados
z_stat, p_valor
```


### Ejemplo 5: Prueba de Hipótesis para Diferencia de Varianzas (Prueba F)

**Contexto**: Supón que eres un analista de datos en una empresa de manufactura y estás comparando dos máquinas que producen la misma pieza. Quieres saber si hay una diferencia significativa en la variabilidad de las dimensiones de las piezas producidas por las dos máquinas. Has recopilado una muestra de las dimensiones producidas por cada máquina.



```python
# Dimensiones de las piezas producidas por las dos máquinas
dimensiones_A = [2.01, 2.00, 2.03, 1.99, 2.02, 2.01, 2.03, 2.00]
dimensiones_B = [2.05, 2.04, 2.02, 2.06, 2.03, 2.05, 2.03, 2.04]

# Calcular las varianzas muestrales
varianza_A = np.var(dimensiones_A, ddof=1)
varianza_B = np.var(dimensiones_B, ddof=1)

# Realizar la prueba F para la diferencia de varianzas
f_stat = varianza_A / varianza_B
df_A = len(dimensiones_A) - 1
df_B = len(dimensiones_B) - 1
p_valor = 2 * min(stats.f.cdf(f_stat, df_A, df_B), 1 - stats.f.cdf(f_stat, df_A, df_B))

# Mostrar los resultados
f_stat, p_valor
```


### Ejemplo 6: Prueba de Hipótesis para Diferencia de Medias (Varianzas Desconocidas y No Asunción de Igualdad)

**Contexto**: Imagina que trabajas en una empresa de retail y quieres comparar los ingresos medios de dos tiendas diferentes en un día particular. No puedes asumir que las varianzas son iguales y no las conoces de antemano. Los ingresos (en miles de dólares) de las dos tiendas son los siguientes:




```python
# Ingresos de las dos tiendas
ingresos_tienda_A = [50, 52, 53, 47, 49, 51, 48, 46, 54, 50]
ingresos_tienda_B = [58, 57, 55, 59, 57, 58, 56, 59, 55, 56]

# Realizar la prueba t para la diferencia de medias (varianzas desconocidas y no asumidas iguales)
t_stat, p_valor = stats.ttest_ind(ingresos_tienda_A, ingresos_tienda_B, equal_var=False)

# Mostrar los resultados
t_stat, p_valor
```

Recuerda que la interpretación del p-valor depende del nivel de significancia que elijas (comúnmente 0.05). Si el p-valor es menor que el nivel de significancia, puedes rechazar la hipótesis nula.

Las muestras pareadas generalmente se usan cuando estás comparando dos conjuntos de observaciones que están intrínsecamente relacionadas de alguna manera. Por ejemplo, medidas repetidas en los mismos sujetos, o medidas pareadas de algún otro modo (como el ejemplo de los tiempos de respuesta antes y después de una optimización).

Por otro lado, las muestras no pareadas se usan cuando las dos muestras son independientes entre sí.


### Ejemplo Hipotético: Tiempos de Respuesta

**Contexto**: Imagina que has implementado una optimización en una aplicación y has medido los tiempos de respuesta antes y después de la optimización en 10 instancias diferentes. Además, has medido los tiempos de respuesta después de la optimización en 10 instancias completamente diferentes.



```python

# Tiempos de respuesta antes de la optimización
tiempos_antes = [210, 200, 215, 220, 205, 210, 200, 215, 220, 205]

# Tiempos de respuesta después de la optimización (instancias pareadas)
tiempos_despues_pareadas = [205, 195, 210, 215, 200, 205, 195, 210, 215, 200]

# Tiempos de respuesta después de la optimización (instancias no pareadas)
tiempos_despues_no_pareadas = [150, 140, 145, 155, 142, 148, 150, 140, 145, 152]

```

```python

# Prueba para muestras pareadas
t_stat_pareadas, p_valor_pareadas = stats.ttest_rel(tiempos_antes, tiempos_despues_pareadas)

# Prueba para muestras no pareadas
t_stat_no_pareadas, p_valor_no_pareadas = stats.ttest_ind(tiempos_antes, tiempos_despues_no_pareadas, equal_var=False)
```

```python


# Mostrar los resultados
t_stat_pareadas, p_valor_pareadas, t_stat_no_pareadas, p_valor_no_pareadas
```
